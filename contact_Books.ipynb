```python
import mysql.connector
```

```python
con=mysql.connector.connect(host="localhost",user="root",password="1234")
```

```python
cursor=con.cursor()
```

```python
cursor.execute("create database project1")
```

```python
cursor.execute('use project1')
```

```python
cursor.execute('create table contacts(id int auto_increment primary key,name varchar(100),phone varchar(15) unique,email varchar(100),address varchar(255))')
```

```python
def add_contact():
    try:
        name=input("Enter name: ")
        phone=input("Enter phone number: ")
        email=input("Enter email: ")
        address=input("Enter address: ")

        sql="insert into contacts (name,phone,email,address) values (%s,%s,%s,%s)"
        val=(name,phone,email,address)
        cursor.execute(sql,val)
        con.commit()
        print("contact added successfully")
    except ValueError:
        
        print("invalid value")
        
```

```python
add_contact()
```

```python
def view_contact():
    cursor.execute("select * from contacts")
    record=cursor.fetchall()
    if record:
        print("\n----- contact list-----")
        for row in record :
            print("ID:",row[0])
            print("name:",row[1])
            print("phone:", row[2])
            print("email:",row[3])
            print("address:",row[4])
            print("-"*30)
        else:
            print("no contacts found")
        con.commit()
```

```python
view_contact()
```

```python
def search_contact():
    name=input("Enter name to search : ")
    cursor.execute("select * from contacts where name=%s",(name,))
    record=cursor.fetchone()
    if record:
        print("\n contact found")
        print("ID:",record[0])
        print("name:",record[1])
        print("phone:", record[2])
        print("email:",record[3])
        print("address:",record[4])
    else:
        print("contact not found")
        
```

```python
search_contact()
```

```python
def update_contact():
    old_phone=input("Enter phone to update: ")
    cursor.execute("select * from contacts where phone=%s",(old_phone,))
    record=cursor.fetchone()
    if record:
        name=input("Enter new name: ")
        new_phone=input("Enter new phone: ")
        email=input("Enter new email: ")
        address=input("Enter new address: ")

        sql="update contacts set name=%s,email=%s,address=%s where phone=%s"
        val=(name,email,address,old_phone)
        cursor.execute(sql,val)
        print("contact updated successfully")
    else:
        print("contact not found")
    con.commit()
```

```python
update_contact()
```

```python
def delete_contace():
    id=int(input())

    sql="delete from contacts where id=%s"
    val=(id,)
    cursor.execute(sql,val)

    if cursor.rowcount>0:
        print("Deleted Successfully")
    else:
        print("Contact not found")
    con.commit()
```

```python
delete_contace()
```

```python
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk
```

```python
window = tk.Tk()
window.title("Contact Book")
window.geometry("500x400")

title=tk.Label(window,text="Contact Book",font=("Arial",20,"bold"))
title.pack(pady=20)

def add_contact_window():
    contact_window=tk.Toplevel(window) 
    contact_window.title("Add Contact")

    contact_window.geometry("400x400")
    #name
    tk.Label(contact_window,text='Name').pack(pady=5)
    name_entry=tk.Entry(contact_window)
    name_entry.pack()
    #phone
    tk.Label(contact_window,text='Phone').pack(pady=5)
    phone_entry=tk.Entry(contact_window)
    phone_entry.pack()
    #Email
    tk.Label(contact_window,text='Email').pack(pady=5)
    Email_entry=tk.Entry(contact_window)
    Email_entry.pack()
    #Address
    tk.Label(contact_window,text='Address').pack(pady=5)
    Address_entry=tk.Entry(contact_window)
    Address_entry.pack()
    def save_contact():
        name= name_entry.get()
        phone=phone_entry.get()
        email=Email_entry.get()
        address=Address_entry.get()
        if name=="" or phone=="": 
            messagebox.showwarning("warning","name and phone are required")
            return
        if not phone.isdigit() or len(phone) !=10:
            messagebox.showwarning("warning","phone number must be 10 digit")
            return
        if not email.endswith("@gmail.com"):
            messagebox.showwarning("warning","email must ends with @gmail.com")
            return
        query="insert into contacts (name,phone,email,address)values(%s,%s,%s,%s)"
        values=(name,phone,email,address)
        cursor.execute(query,values)
        con.commit()
        messagebox.showinfo("Success","contact added successfully")
        contact_window.destroy()

        #save button
    save_btn=tk.Button(contact_window,text="save contact",width=20,command=save_contact)
    save_btn.pack(pady=20)

add_btn=tk.Button(window, text="Add Contact",width=20, command=add_contact_window)
add_btn.pack(pady=10)


view_btn=tk.Button(window, text="View Contact",width=20, command=view_contact_window)
view_btn.pack(pady=10)

search_btn=tk.Button(window, text="search contact",width=20, command=search_contact_window)
search_btn.pack(pady=10)

update_btn=tk.Button(window, text="update Contact",width=20, command=update_contact_window)
update_btn.pack(pady=10)

delete_btn=tk.Button(window, text="delete Contact",width=20, command=delete_contact_window)
delete_btn.pack(pady=10)

window.mainloop()
```

```python
def view_contact_window():
    view_window=tk.Toplevel(window)
    view_window.title("view contact")
    view_window.geometry("700x400")
    tk.Label(view_window,text="all contacts",font=("Arial",18,'bold')).pack(pady=10)
    columns=("ID","name","phone","email","address")
    tree=ttk.Treeview(view_window,columns=columns,show='headings')
    for col in columns:
        tree.heading(col,text=col)
        tree.column(col,width=120)
        tree.pack(fill='both',expand=True,padx=10,pady=10)
        cursor.execute("select * from contacts")
        records=cursor.fetchall()
        for record in records:
            tree.insert("",tk.END,values=record)
```

```python
def search_contact_window():
    search_window=tk.Toplevel(window)
    search_window.title("search contact")
    search_window.geometry("500x350")
    tk.Label(search_window,text="search contacts",font=("Arial",18,'bold')).pack(pady=15)
    tk.Label(search_window,text="Enter phone number:").pack()
    phone_entry=tk.Entry(search_window,width=30)
    phone_entry.pack(pady=5)
    result_label=tk.Label(search_window,text="")
    result_label.pack(pady=10)
    def search():
        phone=phone_entry.get()
        cursor.execute("select * from contacts where phone=%s",(phone,))
        record=cursor.fetchone()
        print(record)
        if record:
            result_label.config(
                text=f"Name: {record[1]}\n"
                     f"Phone: {record[2]}\n"
                     f"Email: {record[3]}\n"
                     f"Address: {record[4]}"
            )
        else:
            result_label.config(text="Contact not found!")

    search_btn=tk.Button(
        search_window,
        text="Search",
        command=search,
        width=15
    ).pack(pady=10)
```

```python
def update_contact_window():
    update_window=tk.Toplevel(window)
    update_window.title("update contact")
    update_window.geometry("500x450")
    tk.Label(update_window,text="update contacts",font=("Arial",18,'bold')).pack(pady=10)
    tk.Label(update_window,text="contact ID").pack()
    id_entry=tk.Entry(update_window) 
    id_entry.pack(pady=5)

    tk.Label(update_window,text="name").pack()
    name_entry=tk.Entry(update_window) 
    name_entry.pack(pady=5)

    tk.Label(update_window,text="phone").pack()
    phone_entry=tk.Entry(update_window) 
    phone_entry.pack(pady=5)

    tk.Label(update_window,text="Email").pack()
    Email_entry=tk.Entry(update_window) 
    Email_entry.pack(pady=5)

    tk.Label(update_window,text="Address").pack()
    Address_entry=tk.Entry(update_window) 
    Address_entry.pack(pady=5)

    def update_contact():
        contact_id=id_entry.get()
        name=name_entry.get()
        phone=phone_entry.get()
        Email=Email_entry.get()
        Address=Address_entry.get()
        cursor.execute("update contacts set name=%s,phone=%s,email=%s,Address=%s where id=%s",(name,phone,Email,Address,contact_id))
        con.commit()
        messagebox.showinfo("sucess","contact update successfully")
    tk.Button(update_window,text="update contact",command=update_contact).pack(pady=15)
```

```python
def delete_contact_window():
    delete_window=tk.Toplevel(window)
    delete_window.title("delete contact")
    delete_window.geometry("400x250")
    tk.Label(delete_window,text="delete contacts",font=("Arial",18,'bold')).pack(pady=15)
    tk.Label(delete_window,text="contact ID").pack()
    id_entry=tk.Entry(delete_window) 
    id_entry.pack(pady=10)
    def delete_contact():
        contact_id=id_entry.get()
        cursor.execute("delete from contacts where id=%s",(contact_id,))
        con.commit()
        messagebox.showinfo("sucess","contact delete successfully")
    tk.Button(delete_window,text="delete contact",command=delete_contact).pack(pady=15)
```
